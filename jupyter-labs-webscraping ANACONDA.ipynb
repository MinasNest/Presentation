````python
import sys

import requests
from bs4 import BeautifulSoup
import re
import unicodedata
import pandas as pd
````

and we will provide some helper functions for you to process web scraped HTML table


````python
def date_time(table_cells):
    """
    This function returns the data and time from the HTML  table cell
    Input: the  element of a table data cell extracts extra row
    """
    return [data_time.strip() for data_time in list(table_cells.strings)][0:2]

def booster_version(table_cells):
    """
    This function returns the booster version from the HTML  table cell 
    Input: the  element of a table data cell extracts extra row
    """
    out=''.join([booster_version for i,booster_version in enumerate( table_cells.strings) if i%2==0][0:-1])
    return out

def landing_status(table_cells):
    """
    This function returns the landing status from the HTML table cell 
    Input: the  element of a table data cell extracts extra row
    """
    out=[i for i in table_cells.strings][0]
    return out


def get_mass(table_cells):
    mass=unicodedata.normalize("NFKD", table_cells.text).strip()
    if mass:
        mass.find("kg")
        new_mass=mass[0:mass.find("kg")+2]
    else:
        new_mass=0
    return new_mass


def extract_column_from_header(row):
    """
    This function returns the landing status from the HTML table cell 
    Input: the  element of a table data cell extracts extra row
    """
    if (row.br):
        row.br.extract()
    if row.a:
        row.a.extract()
    if row.sup:
        row.sup.extract()
        
    colunm_name = ' '.join(row.contents)
    
    # Filter the digit and empty names
    if not(colunm_name.strip().isdigit()):
        colunm_name = colunm_name.strip()
        return colunm_name    
````

````python
extracted_row = 0

# Extract each table (assumed that the correct table is the one with class "wikitable plainrowheaders collapsible")
for table_number, table in enumerate(soup.find_all('table', "wikitable plainrowheaders collapsible")):
    
    # Iterate through all rows in the table
    for rows in table.find_all("tr"):
        flag = False
        
        # Check if the first column contains a flight number (if it's a valid number)
        if rows.th:
            if rows.th.string:
                flight_number = rows.th.string.strip()
                flag = flight_number.isdigit()
        else:
            flag = False
        
        # Extract row data if it's a valid flight number
        row = rows.find_all('td')
        if flag and len(row) > 0:
            extracted_row += 1
            
            # Initialize the launch dictionary for this flight
            launch_dict[flight_number] = {}
            
            # Date and Time from the first column (e.g., row[0])
            datatimelist = date_time(row[0])  # Assuming you have this function defined
            launch_dict[flight_number]['Date'] = datatimelist[0].strip(',')
            launch_dict[flight_number]['Time'] = datatimelist[1]
            
            # Booster version (from the second column, e.g., row[1])
            bv = booster_version(row[1])  # Assuming you have this function defined
            if not bv and row[1].a:
                bv = row[1].a.string
            launch_dict[flight_number]['Version Booster'] = bv
            
            # Launch site (from the third column, e.g., row[2])
            launch_site = row[2].a.string if row[2].a else None
            launch_dict[flight_number]['Launch Site'] = launch_site
            
            # Payload (from the fourth column, e.g., row[3])
            payload = row[3].a.string if row[3].a else None
            launch_dict[flight_number]['Payload'] = payload
            
            # Payload mass (from the fifth column, e.g., row[4])
            payload_mass = get_mass(row[4])  # Assuming you have this function defined
            launch_dict[flight_number]['Payload Mass'] = payload_mass
            
            # Orbit (from the sixth column, e.g., row[5])
            orbit = row[5].a.string if row[5].a else None
            launch_dict[flight_number]['Orbit'] = orbit
            
            # Customer (from the seventh column, e.g., row[6])
            customer = None  # Initialize to None in case no data is found
            if row[6]:
                customer_tag = row[6].find('a')  # Find <a> tag within the cell
                if customer_tag:
                    customer = customer_tag.string  # Get the text of the <a> tag
            launch_dict[flight_number]['Customer'] = customer
            
            # Launch outcome (from the eighth column, e.g., row[7])
            launch_outcome = list(row[7].strings)[0] if row[7].strings else None
            launch_dict[flight_number]['Launch Outcome'] = launch_outcome
            
            # Booster landing (from the ninth column, e.g., row[8])
            booster_landing = landing_status(row[8])  # Assuming you have this function defined
            launch_dict[flight_number]['Booster Landing'] = booster_landing

# Print out the collected launch data (optional)
print(launch_dict)
````
